# FinSight AI — Research Notebook 05: Financial NLP & FinBERT Sentiment Analysis
## Testing Pre-trained ProsusAI/finbert vs. Generic Sentiment Lexicons

```python
import sys
sys.path.append('../backend')
import asyncio
from app.services.sentiment_service import SentimentService

service = SentimentService()

headlines = [
    "NVIDIA reports record quarterly revenue exceeding Wall Street expectations on AI chip demand.",
    "Federal Reserve signals potential rate hikes amid persistent inflation concerns.",
    "Company maintains dividend payout ratio unchanged for the upcoming fiscal quarter.",
    "Automaker recalls 400,000 electric vehicles due to battery overheating risks."
]

async def run_analysis():
    for text in headlines:
        res = await service.analyze(text)
        print(f"Headline: {text}")
        print(f"-> Sentiment: {res['sentiment'].upper()} (Compound: {res['compound_score']:.3f}, Pos: {res['positive_score']:.2f}, Neg: {res['negative_score']:.2f})")
        print("-" * 80)

await run_analysis()
```
